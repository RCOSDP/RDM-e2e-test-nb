## 以下の手順を実施し、ファイルを移動する ・ファイル一覧から「source」フォルダ配下のファイル(1GB)の行をクリックしたまま、Alt キーを押しながらマウスカーソルを移動させる ・ドラッグ状態のファイル(1GB)を「dest」フォルダの行まで移動させ、Alt キーを押したままマウスボタンを離す ※ Mac OS の場合は Option キー

## (「アカウントを切断」が対象ストレージに表示されている場合)すべての「アカウントを切断」をクリックし、「Disconnect」をクリックする



```python
async def _step(page):
    if not has_connections:
        return
    try:
        await expect(page.locator(f'//*[@src="/static/addons/s3/comicon.png"]/../..//*[text() = "アカウントを切断"][1]')).to_be_visible(timeout=30000)
    except:
        print('Ignored')
        traceback.print_exc()
        return
    while True:
        print('Disconnecting...')
        await page.locator(f'//*[@src="/static/addons/s3/comicon.png"]/../..//*[text() = "アカウントを切断"][1]').click()
        await page.locator('.bootbox-confirm .btn-danger').click()
        # 30秒以内に「アカウントを切断」が表示されれば繰り返し
        try:
            # 前のものが消えるまで待つ...
            await asyncio.sleep(3)
            await expect(page.locator(f'//*[@src="/static/addons/s3/comicon.png"]/../..//*[text() = "アカウントを切断"][1]')).to_be_visible(timeout=30000)
        except:
            print('Ignored')
            traceback.print_exc()
            break
    await expect(page.locator(f'//*[@src="/static/addons/s3/comicon.png"]/../..//*[text() = "アカウントを切断"]')).to_have_count(0, timeout=transition_timeout)

await run_pw(_step)
```

## 以下の手順を実施し、ファイルを移動する ・「NII Storage」のファイル一覧からファイル(100MB)の行をクリックしたままマウスカーソルを移動させる ・ドラッグ状態のファイル(100MB)を「Amazon S3」の行まで移動させマウスボタンを離す

以下を確認すること
・ファイル(100MB)の表示行が、「NII Storage」直下から「Amazon S3」の下のツリーに移動すること

```python
filename_100mb = f'{yyyymmdd}_アップロードテスト_100MB.dat'
async def _step(page):
    await page.reload()
    await asyncio.sleep(1)
    await expect(grdm.get_select_file_title_locator(page, filename_100mb)).to_be_visible(timeout=transition_timeout)
    await grdm.get_select_storage_title_locator(page, target_storage_name).click()
    collapsed_toggle = grdm.get_select_folder_toggle_locator(page, target_storage_name, collapsed=True)
    if await collapsed_toggle.count():
        await collapsed_toggle.click()
        await page.wait_for_timeout(1000) 
    await grdm.get_select_file_extension_locator(page, filename_100mb).click()
    source_draggable = grdm.get_select_file_draggable_locator(page, filename_100mb)
    dest_droppable = grdm.get_select_storage_title_locator(page, s3_storage_name).first
    await asyncio.sleep(1)
    await grdm.drag_and_drop(page, source_draggable, dest_droppable)
    await expect(page.get_by_text(re.compile(r"^移動中"))).not_to_be_visible(timeout=transition_timeout * 5)
    await asyncio.sleep(2)

    # NII Storage配下からAmazon S3配下へ移動していること
    await expect(grdm.get_select_storage_title_locator(page, s3_storage_name)).to_be_visible(timeout=transition_timeout)
    await expect(grdm.get_select_file_title_locator(page, filename_100mb)).to_be_visible(timeout=transition_timeout)
await run_pw(_step)
```
