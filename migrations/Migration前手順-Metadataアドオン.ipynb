##  「リポジトリURL・DOIリンク」の「入力」をクリックする

フォルダのURLが表示されること

```python
from urllib.parse import urljoin

async def _step(page):
    await page.locator('//label[contains(text(), "リポジトリURL・DOIリンク")]/../following-sibling::div[1]//*[text() = "入力"]').click()
    value = urljoin(page.url, f'files/dir/{target_storage_id}/TESTMETADATA/')

    await expect(page.locator('//label[contains(text(), "リポジトリURL・DOIリンク")]/../following-sibling::div[1]//input')).to_have_value(value, timeout=10000)
    if not (value.startswith('http://localhost:') or value.startswith('http://192.168.168.167:')):
        return
    # localhost:portはリポジトリURLとして許容されないので、置き換える
    value = value.replace('http://localhost:5000/', 'https://rdm.example.com/')
    value = value.replace('http://localhost:5001/', 'https://rdm.example.com/')
    value = value.replace('http://192.168.168.167:5000/', 'https://rdm.example.com/')
    value = value.replace('http://192.168.168.167:5001/', 'https://rdm.example.com/')
    await page.locator('//label[contains(text(), "リポジトリURL・DOIリンク")]/../following-sibling::div[1]//input').fill(value)

await run_pw(_step)
```

## 「TESTMETADATA」フォルダにファイルをアップロードする

「TESTMETADATA」フォルダ直下にファイルがアップロードされること


```python
import os
import tempfile

async def _step(page):
    upload_path = os.path.join(tempfile.gettempdir(), 'testdata.csv')
    with open(upload_path, 'w') as f:
        f.write('a,b\n1,2\n')
    await expect(grdm.get_select_folder_droppable_locator(page, 'TESTMETADATA')).to_be_visible(timeout=transition_timeout)
    dropzone = grdm.get_select_folder_droppable_xpath('TESTMETADATA')
    await grdm.drop_file(page, dropzone, upload_path)
    await grdm.wait_for_uploaded(page, 'testdata.csv')
    await expect(grdm.get_select_file_title_locator(page, 'testdata.csv')).to_be_visible(timeout=transition_timeout)

await run_pw(_step)
```
